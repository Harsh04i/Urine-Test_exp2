```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv("/content/urinalysis_tests.csv")

if "Unnamed: 0" in df.columns:
    df.drop(columns=["Unnamed: 0"], inplace=True)

df.drop_duplicates(inplace=True)

for col in df.columns:
    if df[col].dtype == object:
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].median())

mapping = {
    "NEGATIVE": 0,
    "TRACE": 0.5,
    "1+": 1,
    "2+": 2,
    "3+": 3,
    "4+": 4,
    "RARE": 1,
    "FEW": 2,
    "MODERATE": 3,
    "MANY": 4,
    "PLENTY": 5,
    "NONE SEEN": 0,
    "LOADED": 10,
    "OCCASIONAL": 1,
    "POSITIVE": 1
}

def convert_value(x):
    if pd.isna(x):
        return np.nan
    x = str(x).strip().upper()
    if "-" in x:
        try:
            a, b = x.split("-")
            return (float(a) + float(b)) / 2
        except:
            pass
    if x in mapping:
        return mapping[x]
    try:
        return float(x)
    except:
        return x

for col in df.columns:
    df[col] = df[col].apply(convert_value)

encoders = {}

for col in df.columns:
    if df[col].dtype == object:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col].astype(str))
        encoders[col] = le

X = df.drop("Diagnosis", axis=1)
y = df["Diagnosis"]

scaler = StandardScaler()
X = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)

processed_df = X.copy()
processed_df["Diagnosis"] = y.values

processed_df.to_csv("/content/urinalysis_tests_preprocessed.csv", index=False)

print(processed_df.head())
print(processed_df.shape)
print("Saved as /content/urinalysis_tests_preprocessed.csv")
```
